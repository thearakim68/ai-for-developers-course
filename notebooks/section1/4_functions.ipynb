### Step 1 · Define and use a function

```
def total(price, quantity):
    return price * quantity

print(total(25, 2))
print(total(10, 3))
```

**Guess** both results before you run.

### Step 2 · Our shop's rule, as a function

```
def can_see(caller, owner):
    return caller == owner

print(can_see("Ana", "Ana"))
print(can_see("Ben", "Ana"))
```

### Step 3 · Calling with a wrong argument (on purpose)
This is the same kind of mistake an AI makes when it invents an argument.

```
total(25, 2, discount=5)
```

Last line: `unexpected keyword argument 'discount'`. The function only accepts what it was built with.

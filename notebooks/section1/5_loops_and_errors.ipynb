### Step 1 · Loop over orders

```
orders = [
    {"id": 10, "owner": "Ana", "item": "keyboard"},
    {"id": 11, "owner": "Ben", "item": "mouse"},
    {"id": 12, "owner": "Ana", "item": "monitor"},
]
for order in orders:
    print(order["id"], order["item"])
```

### Step 2 · Only Ana's orders
**Guess:** how many lines will print?

```
for order in orders:
    if order["owner"] == "Ana":
        print(order["item"])
```

**Change it:** show only Ben's orders.

### Step 3 · Read a traceback (on purpose)
Read it **from the bottom up**: the last line says *what* went wrong, the lines above say *where*.

```
def get_item(order):
    return order["itme"]

get_item(orders[0])
```

Last line: `KeyError: 'itme'`. A typo! Fix `"itme"` to `"item"` and run again.

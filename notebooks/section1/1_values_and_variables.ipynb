# Lesson 1 · Values and variables
A **value** is a piece of data: a number, or some text. A **variable** is a name we stick on a value, so we can use it later. Like a label on a box.

**How to use this notebook**
1. Read the short text above each grey box.
2. **Guess** what will happen. Say it out loud or write it down.
3. Click the ▶ button on the left of the grey box (or press **Shift + Enter**).
4. Compare with your guess. Then try the small **Change it** task.
Nothing here can break your computer. If something looks strange: **Runtime → Restart session**, then run the boxes again from the top.

### Step 1 · Print a value
**Guess:** what will you see?

```
print("Hello, Ana!")
```

### Step 2 · A variable
We put the text in a box called `name`, then print the box.

```
name = "Ana"
print("Hello,", name)
```

**Change it:** put your own name in the quotes, and run again.

### Step 3 · Numbers
Text goes in quotes. Numbers don't. **Guess** the result.

```
price = 25
quantity = 2
print(price * quantity)
```

### Step 4 · Text + number = error (on purpose)
This box will show a red error. That's fine: errors are messages that help you.

```
print("Total: " + 50)
```

Read the **last line** of the error. It says it can only join text to text. The fix: turn the number into text with `str()`.

```
print("Total: " + str(50))
```

In one sentence: what is the difference between `"50"` and `50`?

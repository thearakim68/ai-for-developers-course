### Step 1 · A simple decision
**Guess:** what prints?

```
caller = "Ben"
owner = "Ana"
if caller == owner:
    print("Here is your order")
else:
    print("Order not found")
```

Notice the **4 spaces** before `print`. In Python, spaces show what belongs inside the `if`.

**Change it:** set `caller = "Ana"` and run again.

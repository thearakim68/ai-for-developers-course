### Step 2 · A dictionary: one order

```
order = {"id": 10, "owner": "Ana", "item": "keyboard"}
print(order["item"])
print(order["owner"])
```

### Step 3 · A list of orders
This is how our shop app stores data later in the course.

```
orders = [
    {"id": 10, "owner": "Ana", "item": "keyboard"},
    {"id": 11, "owner": "Ben", "item": "mouse"},
]
print(orders[1]["item"])
```

**Guess first:** whose order is `orders[1]`?
